```python
import pandas as pd
import seaborn as sns
```

```python
df=sns.load_dataset('taxis')
```

```python
df
```

```python
df.groupby(by='pickup_borough')['fare'].agg(['sum','mean'])
```

```python
df.groupby(by='pickup_borough').agg({'fare':['sum','mean'],'tip':['max','min']})
```

```python
df.groupby(by='pickup_borough').agg(total_fare=('fare','sum'),total_tip=('tip','sum'))
```

```python
df.groupby(by='pickup_borough',as_index=False).agg(total_fare=('fare','sum'),total_tip=('tip','sum'))
```

```python
# dot apply method->custom python function applied on the group
def median_fare(group):
    d_median=group[['tip','fare']].median()
    return d_median
```

```python
df.groupby(by='pickup_borough').apply(median_fare,include_groups=False)
```

```python
def range_fare(group):
    range_f=group['fare'].max()-group['fare'].min()
    return range_f
```

```python
df.groupby(by='pickup_borough').apply(range_fare,include_groups=False)
```
