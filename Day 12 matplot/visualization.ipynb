```python
import matplotlib.pyplot as plt
#work as a matlab
```

```python
#create first chart
a=[2, 4, 6, 8, 10]
b=[10, 20,30,40,50]
plt.plot(a,b) #line chart
plt.show()
```

```python
plt.plot(a,b,color='green',marker='o',linestyle='dashed',linewidth='2',markersize='10')
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.plot(a,b,color='red',linestyle='dashed',linewidth='1',markersize='9' , marker='o')
plt.show()
```

```python
plt.figure(figsize=(3,4))
plt.plot(a,b,color='orange',marker='o',linestyle='dashed',linewidth='1',markersize='5')
plt.title("this is plot title")
plt.xlabel("this is xlabel")
plt.ylabel("this is ylabel")
plt.show()
```

```python
#multiple line and legends
x=[1,2,3,4,5]
y1=[37,58,6,48,13]
y2=[38,58,1,9,50]
plt.title("sales data")
plt.xlabel("month")
plt.ylabel("sales")
plt.plot(x,y1,label='sales 2025',color='green')
plt.plot(x,y2,label='sales 2026',color='red')
plt.legend() #to show legend->indexing
plt.show()
```

```python
#bar chart
import matplotlib.pyplot as plt
a=['A','B','C','D','E']
b=[23,45,21,3,12]
plt.figure(figsize=(4,3))
plt.title("Bar chart diagram")
plt.bar(a,b,color='red')
plt.show()
```

```python
import random
```

```python
data= [random.randint(1,10) for i in range(199)]
plt.title("Histogram Example")
plt.hist(data,bins=10,color='green')
plt.show()
```

```python
#pie chart
```

```python
a=['A','B','C','D','E']
b=[23,46,32,67,96]
plt.title('pie chart example')
plt.pie(b,labels=a,autopct='%1.1f%%',startangle=90)
plt.show()
```

```python
#scatter plot
#used to find the relationship between variables
```

```python
a=[2,4,6,8,3]
b=[23,67,32,80,12]
plt.title('scatter plot example')
plt.scatter(a,b,color='orange')
plt.show()
```

```python
#subplot
#used to show multiple chart into single figure
```

```python
a=['A','B','C','D','E']
b=[23,56,89,32,51]

x=[2,4,1,5,8]
y=[46,32,68,6,12]

plt.subplot(1,2,1) #row  column position
plt.title('bar chart')
plt.bar(a,b,color='green')

plt.subplot(1,2,2)
plt.title('scatter chart')
plt.scatter(x,y,color='red')

plt.show()
```

```python
#matplotlib with pandas - real data
import pandas as pd
```

```python
data={
    'months':['jan','feb','mar','april'],
    'sales':[17887,38229,37128,19299]
}
df=pd.DataFrame(data)
```

```python
plt.bar(df['months'],df['sales'],color='pink')
plt.show()
```

```python
plt.bar(df['months'],df['sales'],color='pink')
plt.title('monthly sales')
plt.xlabel('months')
plt.ylabel('sales amount')
plt.savefig('monthly_sales.png')
plt.show()
```
