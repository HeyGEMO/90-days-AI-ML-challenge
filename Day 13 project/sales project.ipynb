```python
import seaborn as sns
import pandas as pd
```

```python
df=pd.read_csv('Sales_Data.csv',encoding='latin1')
```

```python
df.columns
```

```python
df.info('unnamed1')
```

```python
df.drop(columns=['Status','unnamed1'],inplace=True)
```

```python
df.isnull().sum()
```

```python
#df['Amount'].value_counts().head()
```

```python
df.dropna(subset=['Amount'],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.columns
```

```python
ax=sns.countplot(x='Gender',data=df)
```

```python
ax=sns.countplot(x='Gender',data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sales_gen)
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt
ax=sns.countplot(x='Age Group',data=df,hue='Gender')
```

```python
ax=sns.countplot(x='Age Group',data=df,hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_age=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values('Amount',ascending=False)
```

```python
sns.barplot(x='Age Group',y='Amount',data=sales_age)
```

```python
df.columns
```

```python
sales_states=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
```

```python
sns.set(rc={'figure.figsize':(15,5)})
```

```python
sns.barplot(data=sales_states,x='State',y='Orders')
```

```python
sales_state_amt=df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data=sales_state_amt,x='State',y='Amount')
```

```python
df.columns
```

```python
ax=sns.countplot(data=df,x='Marital_Status')
sns.set(rc={'figure.figsize':(3,4)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_marital=df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(data=sales_marital,x='Marital_Status',y='Amount',hue='Gender')
```
