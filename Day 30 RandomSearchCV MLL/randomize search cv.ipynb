```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop('species',axis=1)
y=df['species']
```

```python
x_train , x_test , y_train ,y_test = train_test_split(x,y,test_size=0.2 , random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=5)
```

```python
model_knn.fit(x_train,y_train)
```

```python
model_knn.score(x_train,y_train)
```

```python
from sklearn.svm import SVC
```

```python
model_svm = SVC(gamma='auto') #c=punishment
```

```python
model_svm.fit(x_train,y_train)
```

```python
model_svm.score(x_train,y_train)
```

```python
#now lets use the grid search CV
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_svm),{
    'C':[1,20,30],
    'kernel':['linear','rbf']
},cv=5,return_train_score=False)
```

```python
#not need to do train test split in cross validation
classifier.fit(x,y)
```

```python
classifier.fit(x,y)
classifier.cv_results_
```

```python
results=pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

```python
#Random Search CV
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier_r=RandomizedSearchCV((model_svm),{
    'C':[1,10,20],
    'kernel':['rbf','linear']
},n_iter=4,cv=5,return_train_score=False)
```

```python
classifier_r.fit(x,y)
```

```python
classifier_r.cv_results_
```

```python
results_r=pd.DataFrame(classifier_r.cv_results_)
```
